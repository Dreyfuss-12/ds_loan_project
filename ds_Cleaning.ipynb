```python
#importing pandas library 
import pandas as pd 
#importing the data set
Df = pd.read_csv(r"C:\Users\hp\OneDrive\Desktop\sem 2 school work\decretw mathematics\train_u6lujuX_CVtuZ9i (1).xls")
Df 
```

```python
#removing the NAN values of the loan amount 
Df = Df[Df['LoanAmount'].notna()]
#cleaning the credithistory NAN to unknown
Df['Credit_History'].astype(object).fillna('Unknown')
Df['Credit_History'].isna().sum()
```

```python
#creating the total house hold income
Df['Total_household_income'] = (Df['ApplicantIncome'] + Df['CoapplicantIncome'])
Df.columns
```

```python
#importing regex
import re
#creating a pattern
pattern = r"\+"
#remove the 3+ in dependants
Df['Dependents']= Df['Dependents'].str.replace(pattern,'',regex = True)
Df['Dependents'].value_counts()
```

```python
#changing the coapplicantIncome to in64 
Df['CoapplicantIncome'] = Df['CoapplicantIncome'].astype('int64')
#dataset info 
Df.info() 
#aggregating the applicantincome and the coapplicantincome
Df['Agregate_Income'] = (Df['ApplicantIncome'] + Df['CoapplicantIncome']) / 2
Df.columns
Df 
```

```python
#loan amount term to years 
Df['Loan_Amount_Term(years)'] = Df['Loan_Amount_Term'].apply(lambda x: x //12)
```
